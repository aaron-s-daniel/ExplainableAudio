```python
import os
import urllib.request
import zipfile  # Add this import

# Create the directory
save_path = '/home/mqa887/tmp/audiomnist'
os.makedirs(save_path, exist_ok=True)

# URL of the Audio MNIST dataset
url = "https://github.com/soerenab/AudioMNIST/archive/master.zip"

# Download the dataset
print("Downloading Audio MNIST dataset...")
urllib.request.urlretrieve(url, os.path.join(save_path, "audiomnist.zip"))
print("Download complete.")

# Extract the dataset
print("Extracting files...")
with zipfile.ZipFile(os.path.join(save_path, "audiomnist.zip"), 'r') as zip_ref:
    zip_ref.extractall(save_path)
print("Extraction complete.")

# Remove the zip file
os.remove(os.path.join(save_path, "audiomnist.zip"))

print(f"Audio MNIST dataset is now available at: {save_path}")
```

```python
import os
import numpy as np
import librosa

def load_audiomnist(data_path, sr=16000, duration=1.0):
    audio_path = os.path.join(data_path, 'AudioMNIST-master', 'data')
    
    data = []
    labels = []
    target_length = int(sr * duration)

    # Iterate through folders 01 to 10
    for folder in range(1, 11):
        folder_name = f"{folder:02d}"  # This will give '01', '02', ..., '10'
        folder_path = os.path.join(audio_path, folder_name)
        
        if not os.path.exists(folder_path):
            print(f"Warning: Folder {folder_path} does not exist.")
            continue

        for filename in os.listdir(folder_path):
            if filename.endswith('.wav'):
                file_path = os.path.join(folder_path, filename)
                
                audio, _ = librosa.load(file_path, sr=sr, duration=duration)
                
                # Pad or truncate the audio to the target length
                if len(audio) > target_length:
                    audio = audio[:target_length]
                else:
                    audio = np.pad(audio, (0, max(0, target_length - len(audio))))
                
                data.append(audio)
                labels.append(folder - 1)  # Subtract 1 to get labels 0-9

    return np.array(data), np.array(labels)

# Load the dataset
data_path = '/home/mqa887/tmp/audiomnist'
X, y = load_audiomnist(data_path, duration=1.0)  # Load 1-second clips

print(f"Dataset loaded. Shape of X: {X.shape}, Shape of y: {y.shape}")
```

```python
import os
import numpy as np
import librosa
import json

def load_audiomnist(data_path, sr=16000):
    audio_path = os.path.join(data_path, 'AudioMNIST-master', 'data')
    metadata_path = os.path.join(data_path, 'AudioMNIST-master', 'audioMNIST_meta.json')

    # Load metadata
    with open(metadata_path, 'r') as f:
        metadata = json.load(f)

    data = []
    labels = []

    # Iterate through all audio files
    for filename in os.listdir(audio_path):
        if filename.endswith('.wav'):
            file_path = os.path.join(audio_path, filename)
            
            # Load audio file
            audio, _ = librosa.load(file_path, sr=sr)
            
            # Get label from filename (first character is the digit)
            label = int(filename[0])
            
            data.append(audio)
            labels.append(label)

    return np.array(data), np.array(labels)

# Load the dataset
data_path = '/home/mqa887/tmp/audiomnist'
X, y = load_audiomnist(data_path)

print(f"Dataset loaded. Shape of X: {X.shape}, Shape of y: {y.shape}")
```
